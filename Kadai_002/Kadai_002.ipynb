```python
import pandas as pd
df = pd.DataFrame({
    '日付':
    ['2023-05-17', '2023-05-18', '2023-05-19', '2023-05-20', '2023-05-21'],
    '社員名':
    ['山田', '佐藤', '鈴木', '田中', '高橋'],
    '売上':
    [100, 200, 150, 300, 250],
    '部門':
    ['メーカー', '代理店', 'メーカー', '商社', '代理店'],
})

df['平均売上'] = df['売上'].mean() #売上の平均を平均売上列へ
df['平均売上'] = pd.to_numeric(df['平均売上'])
goal = df['売上'].mean() + 50

def performance(rank):
  result = ''

  if rank >= goal:
    result = 'A'
  elif rank < goal - 50:
    result = 'C'
  else:
    result = 'B'
  return result

df['業績ランク'] = df['売上'].apply(performance)

#クラス→ExcelWriter 	DataFrameオブジェクトをExcelファイルに書き込む
writer = pd.ExcelWriter('業績.xlsx')
df.to_excel(writer, index=False)
writer.close()
```

```python
import pandas as pd
import openpyxl
from openpyxl.styles import PatternFill

df1 = pd.read_excel('2022_年間売上表.xlsx', sheet_name='Sheet1')
df2 = pd.read_excel('2023_年間売上表.xlsx', sheet_name='Sheet1')

df = pd.DataFrame()
df = pd.concat([df1, df2], ignore_index=True)

result = df.groupby(['商品', '売上年'] )['金額（千円）'].sum().reset_index()
writer = pd.ExcelWriter('売上集計表.xlsx')
result.to_excel(writer, sheet_name='Sheet1', index=False)
writer.close()

workbook = openpyxl.load_workbook('売上集計表.xlsx')
sheet = workbook.active

fill = PatternFill(fill_type='solid', fgColor='F2F2F2')
for col in range(1, 4):
    cell = sheet.cell(row=1, column=col)
    cell.fill = fill

workbook.save('売上集計表.xlsx')
```

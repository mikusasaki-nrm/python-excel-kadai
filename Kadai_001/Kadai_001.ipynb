```python
import datetime
today = datetime.date.today()
import openpyxl
workbook = openpyxl.Workbook()
ws = workbook.active

#会社情報部分
ws['A2'] = '請求書'
ws['A4'] = '株式会社ABC'
ws['A5'] = '〒101-0022 東京都千代田区神田練塀町300'
ws['A6'] = 'TEL:03-1234-5678 FAX:03-1234-5678'
ws['A7'] = '担当者名:鈴木一郎 様'

#日付と請求書番号
ws['E4'] = 'No.'
ws['F4'] = '0001'
ws['E5'] = '日付'
ws['F5'] = today.strftime("%Y/%m/%d")

#請求内訳
header = ['商品名', '数量', '価格','金額']
ws.append(header)
data = [
    ['商品A', 2, 10000],
    ['商品B', 1, 15000],
]
total = 0
for row in data:  #dataの中身を取り出してrowに入れながら繰り返す
  price = row[1] * row[2]
  total = total + price
  row.append(price)
  ws.append(row)
ws['D11'] = total

#請求合計
amount = [
    ['合計', '', '', total],
    ['消費税', '', '', total * 0.1],
    ['税込合計', '', '', total + total * 0.1]
]
for row in amount:  #amountの中身を取り出してrowに入れながら繰り返す
  ws.append(row)

# 指定した行と列に新しい行と列を挿入
ws.insert_rows(8, 2)
ws.insert_rows(14, 1)
ws.insert_cols(1, 1)
workbook.save('請求書.xlsx')
```
